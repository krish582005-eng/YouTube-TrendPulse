```python
import pandas as pd

df = pd.read_csv("../data/raw/INvideos.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))
```

```python
print(df["video_id"].head(10))
print("Unique video IDs:", df["video_id"].nunique())
```

```python
clean_df = df.copy()
```

```python
duplicate_count = clean_df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
clean_df = clean_df.drop_duplicates().reset_index(drop=True)
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
numeric_columns = [
    "views",
    "likes",
    "dislikes",
    "comment_count"
]
```

```python
for column in numeric_columns:
    clean_df[column] = pd.to_numeric(
        clean_df[column],
        errors="coerce"
    )
```

```python
clean_df["trending_date"] = pd.to_datetime(
    clean_df["trending_date"],
    format="%y.%d.%m",
    errors="coerce"
)
```

```python
clean_df["publish_time"] = pd.to_datetime(
    clean_df["publish_time"],
    errors="coerce",
    utc=True
)
```

```python
print(
    "Invalid trending dates:",
    clean_df["trending_date"].isna().sum()
)

print(
    "Invalid publish times:",
    clean_df["publish_time"].isna().sum()
)
```

```python
missing_values = clean_df.isna().sum()

print(
    missing_values[missing_values > 0]
)
```

```python
print("Negative views:",
      (clean_df["views"] < 0).sum())

print("Negative likes:",
      (clean_df["likes"] < 0).sum())

print("Negative dislikes:",
      (clean_df["dislikes"] < 0).sum())

print("Negative comments:",
      (clean_df["comment_count"] < 0).sum())
```

```python
clean_df["publish_date"] = (
    clean_df["publish_time"]
    .dt.tz_localize(None)
    .dt.normalize()
)
```

```python
clean_df["time_to_trend_days"] = (
    clean_df["trending_date"]
    - clean_df["publish_date"]
).dt.days
```

```python
print(
    "Negative Time-to-Trend:",
    (clean_df["time_to_trend_days"] < 0).sum()
)
```

```python
clean_df = clean_df[
    clean_df["views"] >= 0
].copy()
```

```python
query = """
CREATE VIEW IF NOT EXISTS category_performance AS

SELECT
    category_id,
    COUNT(DISTINCT video_id) AS unique_videos,
    SUM(views) AS total_views,
    SUM(likes) AS total_likes,
    SUM(comment_count) AS total_comments,
    AVG(views) AS average_views

FROM youtube_trends

GROUP BY category_id

ORDER BY total_views DESC
"""

conn.execute(query)
conn.commit()
```

```python
velocity_df["growth_score"] = velocity_df["velocity_percentile"]

velocity_df["engagement_score"] = velocity_df["engagement_percentile"]

velocity_df["emerging_trend_score"] = (
    0.6 * velocity_df["growth_score"]
    + 0.4 * velocity_df["engagement_score"]
)
```

```python
velocity_df["emerging_status"] = pd.cut(
    velocity_df["emerging_trend_score"],
    bins=[-1, 40, 60, 80, 100],
    labels=["Normal", "Growing", "Emerging", "Hot"]
)
```

```python
velocity_df["trend_alert"] = velocity_df["emerging_status"].isin(
    ["Emerging", "Hot"]
)
```
